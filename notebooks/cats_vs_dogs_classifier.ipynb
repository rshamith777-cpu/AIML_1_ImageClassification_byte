### Step 1: Environment Setup & Imports
We import NumPy for matrix operations, Pillow for image parsing, and scikit-learn for preprocessing, pipeline assembly, and evaluation metrics.

### Step 2: Dataset Verification
We inspect our sample dataset directory and verify class balance.

```python
sample_dir = '../outputs/sample_inferences'
files = [f for f in os.listdir(sample_dir) if f.endswith('.jpg')]
cats = [f for f in files if 'cat' in f]
dogs = [f for f in files if 'dog' in f]
print(f'Total sample inference images: {len(files)}')
print(f'Cat samples ({len(cats)}): {cats}')
print(f'Dog samples ({len(dogs)}): {dogs}')
```

### Step 3: Feature Extraction Demo
We test feature extraction on a sample image. The vector includes normalized 32x32 spatial color pixels, channel moments (mean, std), and 16-bin color histograms.

```python
test_img_path = os.path.join(sample_dir, 'cat_01.jpg')
features = extract_image_features(test_img_path)
print(f'Extracted feature vector shape: {features.shape}')
print(f'Sample feature values (first 10): {np.round(features[:10], 4)}')
```

### Step 4: Model Pipeline Architecture
We assemble our machine learning pipeline combining standard scaling and regularized logistic classification.

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(C=1.0, max_iter=1000, random_state=42))
])
print('Pipeline structure: StandardScaler -> LogisticRegression(C=1.0, max_iter=1000)')
```
